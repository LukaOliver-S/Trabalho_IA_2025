```python
import matplotlib.pyplot as plt
from fuzzy_controller import FuzzySimple

# Instantiate controller
flc = FuzzySimple()

# Plot MFs: entrada (dist_front) e saída (v)
fig, axs = plt.subplots(1, 2, figsize=(12, 4))
flc.dist_front.view(ax=axs[0])
axs[0].set_title('dist_front MFs')

# v é a saída única (vx = vy)
flc.v.view(ax=axs[1])
axs[1].set_title('v (vx = vy) MFs')

plt.tight_layout()
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

# Instanciar o controlador simples ou o TSK original
flc = FuzzySimple()   # ou FuzzySimple() se estiver usando a versão escalar

lateral = 0.0  # centered
p_vals = np.linspace(0.01, 1.0, 200)
vx_vals, vy_vals = [], []

for d in p_vals:
    vx, vy = flc.compute_velocity(d, lateral)
    vx_vals.append(vx)
    vy_vals.append(vy)

plt.figure(figsize=(7, 4))
plt.plot(p_vals, vx_vals, label='vx (m/s)')
plt.plot(p_vals, vy_vals, label='vy (m/s)')
plt.xlabel('front distance (m)')
plt.ylabel('velocity (m/s)')
plt.title('vx and vy vs front distance (lateral=0.0 m)')
plt.grid(True)
plt.legend()
plt.show()
```

```python
%matplotlib inline
import pandas as pd
from pathlib import Path
import matplotlib.pyplot as plt
import seaborn as sns
sns.set(style="whitegrid")

folder = Path("verifying_fuzzy")
csvs = sorted(folder.glob("*.csv"))
print("Folder:", folder.resolve(), "CSV files:", len(csvs))
for f in csvs:
    print(" -", f.name)
    df = pd.read_csv(f)
    print("  rows:", len(df), "columns:", list(df.columns))
    # show head
    display(df.head())

    # guess dist column and velocity-like columns
    def find_col(df, keys):
        lc = [c.lower() for c in df.columns]
        for k in keys:
            for i,c in enumerate(lc):
                if k in c:
                    return df.columns[i]
        return None

    dist_col = find_col(df, ("dist","distance","front","range"))
    vcols = [c for c in df.columns if find_col(pd.DataFrame(columns=[c]), ("vx","vy","v_","vel","desired","cap","v")) is not None]

    if dist_col and vcols:
        plt.figure(figsize=(8,4))
        for c in vcols:
            plt.plot(df[dist_col], df[c], label=c)
        plt.xlabel(dist_col); plt.legend(); plt.title(f.name); plt.show()
    elif df.select_dtypes("number").shape[1] >= 2:
        # fallback: numeric columns vs index + corr heatmap
        plt.figure(figsize=(8,4))
        for c in df.select_dtypes("number").columns:
            plt.plot(df.index, df[c], label=c)
        plt.legend(); plt.title(f"{f.name} (numeric cols vs index)"); plt.show()
        plt.figure(figsize=(6,5)); sns.heatmap(df.select_dtypes("number").corr(), annot=True, fmt=".2f"); plt.show()
    else:
        print("  No numeric columns to plot.")
```
